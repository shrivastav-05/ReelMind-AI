```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
import time
```

```python
driver=webdriver.Chrome()
driver.get("https://www.workindia.in/")
time.sleep(5)
```

```python
input("login manually and press ENTER here to continue...")
job_input= driver.find_element(By.XPATH,"//input[@placeholder='Eg. Back Office job']")
job_input.click()
job_input.send_keys("Software Developer")
time.sleep(2)
job_input.send_keys(Keys.RETURN)

city_input=driver.find_elements(By.XPATH,"//Pick Your City")
city_input.click()
city_input.send_keys("Delhi/NCR,DL")
#time.sleep(3)
city_input.send_keys(keys.RETURN)

search_btn=driver.find_element(By.XPATH,"//button")
search_btn.click()

time.sleep(6)
```
