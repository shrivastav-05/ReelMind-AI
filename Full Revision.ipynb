```python
#Slicing

arr_03=np.array([1,2,3,4,5,6,7])
#                0,1,2,3,4,5,6

print(arr_03)
print()
print("2 to 5:",arr_03[1:3])
print("2 to End :",arr_03[1:])
print("start to 5:",arr_03[:5])
print("stop:",arr_03[::2])
print("stop:",arr_03[1:8:2])
```

```python
#Iteration:_
```

```python
arr_05=np.array([9,8,7,6,5,4])
print(arr_05)
print()
for i in arr_05:
    print(i)
```

```python
arr_06=np.array([[1,2,3,4],[1,2,3,4]])
print(arr_06)
print()

for j in arr_06:
    print(j)

print()

for k in arr_06:
    for l in k:
        print(l)
```

```python
arr_07=np.array([[[1,2,3,4],[4,5,6,7]]])
print(arr_07)
print(arr_07.ndim)
print()

for i in np.nditer(arr_07): #nditer:- for using iteration
    print(i)
```

```python
#ndenu:- iteration with index.
arr_07=np.array([[[1,2,3,4],[4,5,6,7]]])
print(arr_07)
print(arr_07.ndim)
print()

for i in np.ndenumerate(arr_07):
    print(i)
```

```python
#Join and Split Array

arr_10=np.array([1,2,3,4])
arr_11=np.array([9,8,7,6])
ar=np.concatenate((arr_10,arr_11))
print(ar)
```

```python
arr_12=np.array([[11,12],[13,14]])
arr_13=np.array([[9,8],[7,6]])
ar_new=np.concatenate((arr_12,arr_13),axis=1)
print(arr_12)
print()
print(arr_13)
print()
print(ar_new)
```

```python
arr_14=np.array([1,2,3,4])
arr_15=np.array([9,8,7,6])

ar_new2=np.stack((arr_14,arr_15),axis=1)
ar_new3=np.hstack((arr_14,arr_15)) #row ke along
ar_new4=np.vstack((arr_14,arr_15)) #colums ke along
ar_new5=np.dstack((arr_14,arr_15)) #heights ke along

print(ar_new2)
print()
print(ar_new3)
print()
print(ar_new4)
print()
print(ar_new5)
```
