```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
aqs = pd.read_csv("aqs_data.csv")
purple = pd.read_csv("PurpleAir_data.csv")

aqs['Date'] = pd.to_datetime(aqs['Date'], format="mixed", dayfirst=True)
purple['Date'] = pd.to_datetime(purple['Date'], format="mixed", dayfirst=True)

print(aqs.head())
print(purple.head())
```

```python
df = pd.read_csv("sales.csv")

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Revenue'] = df['UnitsSold'] * df['UnitPrice'] * (1 - df['Discount'])

print(df.head())
```

```python
store_rev = df.groupby(['Date','StoreID'])['Revenue'].sum().reset_index()

top_store_per_day = store_rev.loc[store_rev.groupby('Date')['Revenue'].idxmax()]

print(top_store_per_day)
```

```python
cat_rev = df.groupby(['Date','ProductCategory'])['Revenue'].sum().reset_index()

cat_rev['PctChange'] = cat_rev.groupby('ProductCategory')['Revenue'].pct_change() * 100

print(cat_rev)
```

```python
df['RevenuePerUnit'] = df['Revenue'] / df['UnitsSold']

avg_rev_per_unit = df.groupby('StoreID')['RevenuePerUnit'].mean()

print(avg_rev_per_unit)
```

```python
top_store = avg_rev_per_unit.idxmax()

print("Store with highest avg revenue per unit:", top_store)
```

```python
category_rev = df.groupby('ProductCategory')['Revenue'].sum().sort_values(ascending=False)

cum_percent = category_rev.cumsum() / category_rev.sum()
top_categories = category_rev.head(2)

print(top_categories)
```
